```python
print(np.sum(arr2))
print(np.max(arr2))
print(np.min(arr2))
print(np.mean(arr2))
print(np.median(arr2))
print(np.sum(arr2,axis=0))
print(np.sum(arr2,axis=1))
print(np.max(arr2,axis=0))
print(np.max(arr2,axis=1))
print(np.min(arr2,axis=0))
print(np.min(arr2,axis=1))
print(np.mean(arr2,axis=0))
print(np.mean(arr2,axis=1))
print(np.median(arr2,axis=0))
print(np.median(arr2,axis=1))
```

```python
students_subjectwise_marks=np.array([[70,80,90],[68,56,78],[77,89,76],[65,79,90]])
print(f"Average marks of each subject={np.mean(students_subjectwise_marks,axis=0)}")
print(f"Highest marks in each subject={np.max(students_subjectwise_marks,axis=0)}")
print(f"All marks greater than 65:{students_subjectwise_marks[students_subjectwise_marks>65]}")
print(f"Overall Average mark of all subjects={np.mean(students_subjectwise_marks)}")
```
